```python
import cv2
import numpy as np
import os
from pathlib import Path

# Configuration
input_dir = Path("/home/docker/workspace/workspace/src/real-time-3D-tracking/test/inputs/videos")    # Directory containing JPG images
output_dir = Path("/home/docker/workspace/workspace/src/real-time-3D-tracking/scripts")  # Directory to save processed images
output_dir.mkdir(exist_ok=True)

# HSV threshold (example: detect red-ish colors)

lower_hsv = np.array([
        58,
        63,
        109
    ])      # Lower bound of HSV threshold
upper_hsv = np.array([
        83,
        101,
        210
    ])     # Upper bound of HSV threshold

# Process each JPG image in the input directory
for img_path in input_dir.glob("*.jpg"):
    # Load image
    img_bgr = cv2.imread(str(img_path))
    if img_bgr is None:
        print(f"Failed to load {img_path}")
        continue

    # Convert to HSV
    img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)

    # Apply threshold
    mask = cv2.inRange(img_hsv, lower_hsv, upper_hsv)

    # Convert original image to grayscale and back to BGR
    gray = cv2.cvtColor(cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY), cv2.COLOR_GRAY2BGR)

    # Create red overlay for masked areas
    red_overlay = np.zeros_like(img_bgr)
    red_overlay[:, :] = [0, 0, 255]  # Red in BGR

    # Combine grayscale image with red overlay where mask is active
    result = np.where(mask[:, :, np.newaxis] == 255, red_overlay, gray)

    # Save result
    output_path = output_dir / img_path.name
    cv2.imwrite(str(output_path), result)

    print(f"Saved: {output_path}")
```
